## Data Reading
**Import libraries**

```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

**Reasding CSV data**

**Trip Type Data**

```python
df_trip_type = spark.read.format('csv')\
                    .option('inferSchmea',True)\
                        .option('header',True)\
                            .load("abfss://bronze@nyctaxistorageanand.dfs.core.windows.net/trip_type")
                            
```

```python
df_trip_type.display()
```

**Trip Zone Data**

```python
df_trip_zone = spark.read.format('csv')\
                    .option('inferSchmea',True)\
                        .option('header',True)\
                            .load("abfss://bronze@nyctaxistorageanand.dfs.core.windows.net/trip_zone")
df_trip_zone.display()
```

**Trip Data (Parquet Files)**

***Using RecursiveFileLookup function we can read the data withion the folders/or hierarchical folder data reading***

**We can difine the schema as well insetad of spark's inferSchema to run. Using two types DDL and StructType**

```python
myschema = '''
                VendorID BIGINT,
                lpep_pickup_datetime TIMESTAMP,
                lpep_dropoff_datetime TIMESTAMP,
                store_and_fwd_flag STRING,
                RatecodeID BIGINT,
                PULocationID BIGINT,
                DOLocationID BIGINT,
                passenger_count BIGINT,
                trip_distance DOUBLE,
                fare_amount DOUBLE,
                extra DOUBLE,
                mta_tax DOUBLE,
                tip_amount DOUBLE,
                tolls_amount DOUBLE,
                ehail_fee DOUBLE,
                improvement_surcharge DOUBLE,
                total_amount DOUBLE,
                payment_type BIGINT,
                trip_type BIGINT,
                congestion_surcharge DOUBLE
            '''
```

```python
df_trip = spark.read.format('parquet')\
                    .schema(myschema)\
                        .option('header',True)\
                            .option('recursiveFileLookup',True)\
                                .load("abfss://bronze@nyctaxistorageanand.dfs.core.windows.net/trips2023data/")
df_trip.display()
```

```python
df_trip.count()
```

# **Data Transformation**

**Trip Data**

```python
df_trip_type=df_trip_type.withColumnRenamed('description','trip_description')
df_trip_type.display()
```

**Write into storageaccount silver layer**

```python
df_trip_type.write.format('parquet')\
    .mode('append')\
        .option('path',"abfss://silver@nyctaxistorageanand.dfs.core.windows.net/trip_type")\
            .save()
```

**Trip Zone**

```python
df_trip_zone.display()
```

```python
df_trip_zone= df_trip_zone.withColumn('zone1', get(split(col('Zone'),'/'), 0))\
    .withColumn('zone2', get(split(col('Zone'),'/'), 1))

df_trip_zone.display()
```

```python
df_trip_zone.write.format('parquet')\
    .mode('append')\
        .option('path',"abfss://silver@nyctaxistorageanand.dfs.core.windows.net/trip_zone")\
            .save()
```

**Trip Data**

```python
df_trip=df_trip.withColumn('trip_date',to_date('lpep_pickup_datetime'))\
    .withColumn('trip_year',year('lpep_pickup_datetime'))\
        .withColumn('trip_month',month('lpep_pickup_datetime'))

df_trip.display()
```

```python
df_trip = df_trip.select('VendorID','PULocationID','DOLocationID','fare_amount','total_amount')
df_trip.display()
```

```python
df_trip.write.format('parquet')\
    .mode('append')\
        .option('path',"abfss://silver@nyctaxistorageanand.dfs.core.windows.net/tripsdata")\
            .save()
```
